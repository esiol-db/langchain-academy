````python
from typing import TypedDict
from langgraph.graph import START, END, StateGraph
````

````python
class AgentState(TypedDict):
  graph_state: str
````

````python
def node1(state: AgentState) -> AgentState:
  state["graph_state"] = state["graph_state"] + "I am "
  return state

def node2(state: AgentState) -> AgentState:
  state["graph_state"] = state["graph_state"] + "sad!"
  return state

def node3(state: AgentState) -> AgentState:
  state["graph_state"] = state["graph_state"] + "happy!"
  return state
````

````python
import random
from typing import Literal

def condition_edge(state: AgentState) -> Literal['node2', 'node3']:
  if random.random() > 0.5:
    return 'node2'
  return 'node3'
````

````python
graph = StateGraph(AgentState)

graph.add_node('node1', node1)
graph.add_node('node2', node2)
graph.add_node('node3', node3)

graph.add_edge(START, 'node1')
graph.add_conditional_edges('node1', condition_edge)
graph.add_edge('node2', END)
graph.add_edge('node3', END)

app = graph.compile()
app
````

````python
app.invoke({'graph_state': "Hi, this is Lance."})
````

````python
from typing import TypedDict, Annotated
from operator import add
from langgraph.graph import START, END, StateGraph

class AgentState(TypedDict):
  graph_state: Annotated[str, add]
````

````python
def node1(state: AgentState) -> AgentState:
  state["graph_state"] = " I am"
  return state

def node2(state: AgentState) -> AgentState:
  state["graph_state"] = " sad!"
  return state

def node3(state: AgentState) -> AgentState:
  state["graph_state"] = " happy!"
  return state
````

````python
import random
from typing import Literal

def condition_edge(state: AgentState) -> Literal['node2', 'node3']:
  if random.random() > 0.5:
    return 'node2'
  return 'node3'
````

````python
graph = StateGraph(AgentState)

graph.add_node('node1', node1)
graph.add_node('node2', node2)
graph.add_node('node3', node3)

graph.add_edge(START, 'node1')
graph.add_conditional_edges('node1', condition_edge)
graph.add_edge('node2', END)
graph.add_edge('node3', END)

app = graph.compile()
app
````

````python
app.invoke({'graph_state': "Hi, this is Lance."})
````
